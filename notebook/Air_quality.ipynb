```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('/content/drive/My Drive/Air Quality Prediction/aqidaily_fiveyears.csv')
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Filter 2020 data
df_2020 = df[df['Date'].dt.year == 2020].copy()
df_2020['Month'] = df_2020['Date'].dt.strftime('%b')

# Pollutants
pollutants = ['CO', 'Ozone', 'PM10', 'PM25', 'NO2']

# Convert to numeric
for pol in pollutants:
    if pol in df_2020.columns:
        df_2020[pol] = pd.to_numeric(df_2020[pol], errors='coerce')

# Monthly averages
monthly_means = df_2020.groupby('Month')[pollutants].mean()
ordered_months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
monthly_means = monthly_means.reindex(ordered_months)

# Plot with enhancements
plt.figure(figsize=(12, 7))
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd'] # Custom colors

for i, pol in enumerate(pollutants):
    if pol in monthly_means and not monthly_means[pol].isnull().all():
        plt.plot(monthly_means.index, monthly_means[pol], marker='o', label=pol,
                 linewidth=2.5, markersize=8, color=colors[i])

        # Add data labels on each point
        for x, y in zip(monthly_means.index, monthly_means[pol]):
            if not np.isnan(y):
                plt.text(x, y + 2, f'{y:.1f}', ha='center', fontsize=8, color=colors[i])

        # Highlight maximum value
        max_val = monthly_means[pol].max()
        max_month = monthly_means[pol].idxmax()
        plt.scatter(max_month, max_val, s=200, color='red', zorder=5, edgecolor='black', linewidth=2)
        plt.text(max_month, max_val + 10, f'Max: {max_val:.1f}', ha='center', fontsize=9,
                 fontweight='bold', color='red')

plt.xlabel('Month in 2020', fontsize=13, fontweight='bold')
plt.ylabel('Average Pollutant Level (µg/m³)', fontsize=13, fontweight='bold')
plt.title('Monthly Air Quality Pollutant Levels - 2020', fontsize=16, fontweight='bold')
plt.legend(fontsize=11, loc='upper left')
plt.grid(axis='y', alpha=0.4, linestyle='--')
plt.ylim(0,160)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('/content/drive/My Drive/Air Quality Prediction/aqidaily_fiveyears.csv')
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Filter only the 5 required years
df = df[df['Date'].dt.year.isin([2020, 2021, 2022, 2023, 2025])].copy()
df['Year'] = df['Date'].dt.year

# Pollutants
pollutants = ['CO', 'Ozone', 'PM10', 'PM25', 'NO2']

# Convert pollutants to numeric
for pol in pollutants:
    if pol in df.columns:
        df[pol] = pd.to_numeric(df[pol], errors='coerce')

# Yearly averages
yearly_means = df.groupby('Year')[pollutants].mean()

# Plot setup
plt.figure(figsize=(12, 7))
bar_width = 0.15
x = np.arange(len(yearly_means.index))
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

# Create grouped bars for each pollutant
for i, pol in enumerate(pollutants):
    if pol in yearly_means.columns:
        plt.bar(x + i * bar_width, yearly_means[pol], width=bar_width,
                label=pol, color=colors[i])

        # Add data labels
        for xi, val in zip(x, yearly_means[pol]):
            if not np.isnan(val):
                plt.text(xi + i * bar_width, val + 2, f'{val:.1f}',
                         ha='center', fontsize=8, color='black', rotation=90)

# Graph formatting
plt.xlabel('Year', fontsize=13, fontweight='bold')
plt.ylabel('Average Pollutant Level (µg/m³)', fontsize=13, fontweight='bold')
plt.title('Average Air Quality Pollutant Levels (2020–2025)', fontsize=16, fontweight='bold')
plt.xticks(x + bar_width * 2, yearly_means.index)
plt.legend(fontsize=11, loc='upper right')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.ylim(0, 160)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# --- 1. Load Data and Initial Processing ---
# Load the dataset
df = pd.read_csv('/content/drive/My Drive/Air Quality Prediction/aqidaily_fiveyears.csv')


# Convert 'Date' to datetime and extract 'Year'
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y')
df['Year'] = df['Date'].dt.year

# --- 2. Data Cleaning and Preparation ---
# Define pollutant columns
pollutant_cols = ['CO', 'Ozone', 'PM10', 'PM25', 'NO2']

# Convert object columns to numeric, coercing non-numeric values (like '.') to NaN
for col in pollutant_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Rename 'PM25' column for consistency and update the pollutant list
df = df.rename(columns={'PM25': 'PM2.5'})
pollutant_cols = ['CO', 'Ozone', 'PM10', 'PM2.5', 'NO2']

# --- 3. Aggregation and Reshaping ---
# Calculate the mean value for each pollutant per year
# NaN values (resulting from 'errors=coerce') are automatically ignored in the mean calculation
yearly_avg_pollutants = df.groupby('Year')[pollutant_cols].mean().reset_index()

# **FIX:** Melt the DataFrame from wide to long format.
# This creates the 'Pollutant' and 'Average_Value' columns required for the pivot table.
df_melted = yearly_avg_pollutants.melt(
    id_vars='Year',
    value_vars=pollutant_cols,
    var_name='Pollutant',
    value_name='Average_Value'
)

# Pivot the melted data to a matrix format for the heatmap: Year on index, Pollutant on columns
heatmap_data = df_melted.pivot_table(
    index='Year',
    columns='Pollutant',
    values='Average_Value'
)

# --- 4. Visualization (Heatmap) ---
# Set the style for the plot
sns.set_theme(style="whitegrid")

# Create the figure and axes
plt.figure(figsize=(10, 6))

# Generate the heatmap
sns.heatmap(
    heatmap_data,
    annot=True, # Display the value in each cell
    cmap="YlOrRd", # Attractive color map (Yellow-Orange-Red)
    fmt=".1f", # Format annotations to one decimal place
    linewidths=.5, # Add lines between cells for clarity
    cbar_kws={'label': 'Average Pollutant Value'} # Label the color bar
)

# Set the title and labels
plt.title('Average Pollutant Values by Year (2020-2024)', fontsize=16)
plt.ylabel('Year', fontsize=12)
plt.xlabel('Pollutant', fontsize=12)

# Ensure y-axis labels (Years) are displayed horizontally
plt.yticks(rotation=0)

# Save the plot and show it
plt.savefig('average_pollutant_heatmap_by_year.png', bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor

# Load and preprocess dataset
df = pd.read_csv('/content/drive/My Drive/Air Quality Prediction/aqidaily_fiveyears.csv')
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df = df.dropna(subset=['Date'])

# Extract year and month (numeric) features for prediction
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month

# Target pollutants
pollutants = ['CO', 'Ozone', 'PM10', 'PM25', 'NO2']

# Convert all pollutant columns to numeric
for pol in pollutants:
    df[pol] = pd.to_numeric(df[pol], errors='coerce')

# Drop rows with missing pollutant data
df = df.dropna(subset=pollutants + ['Year', 'Month'])

# Features: Year and Month
X = df[['Year', 'Month']]

# Will store min and max prediction values for each pollutant
min_preds = []
max_preds = []

# Predict pollutant values for every month in 2026
future_year = 2026
future_months = np.arange(1, 13)
X_future = pd.DataFrame({'Year': [future_year]*12, 'Month': future_months})

for pol in pollutants:
    y = df[pol]

    # Train/test split (for robustness; optional)
    X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

    # Train model
    model = RandomForestRegressor(n_estimators=100, random_state=42)
    model.fit(X_train, y_train)

    # Predict 2026 monthly values
    preds = model.predict(X_future)

    # Store min and max predicted values for pollutant
    min_preds.append(np.min(preds))
    max_preds.append(np.max(preds))

# Plotting min and max predicted values side by side as bars for each pollutant
bar_width = 0.35
index = np.arange(len(pollutants))

fig, ax = plt.subplots(figsize=(12, 7))

bars_min = ax.bar(index, min_preds, bar_width, label='Min Predicted', color='darkblue')
bars_max = ax.bar(index + bar_width, max_preds, bar_width, label='Max Predicted', color='red')

ax.set_xlabel('Pollutants', fontsize=13, fontweight='bold')
ax.set_ylabel('Predicted Level (µg/m³)', fontsize=13, fontweight='bold')
ax.set_title('Predicted Minimum and Maximum Pollutant Levels in 2026', fontsize=16, fontweight='bold')
ax.set_xticks(index + bar_width / 2)
ax.set_xticklabels(pollutants)
ax.legend(fontsize=11)

# Add value labels above bars (bold)
def add_value_labels(bars):
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height:.1f}',
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 5),  # 5 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=10, fontweight='bold')

add_value_labels(bars_min)
add_value_labels(bars_max)

plt.tight_layout()
plt.show()
```
